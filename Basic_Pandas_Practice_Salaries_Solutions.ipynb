Import pandas

```python
import pandas as pd
```

Read this csv into a dataframe called df

```python
# Read dataset from csv file 
# If you save the Jupyter Notebook in the same location as your dataset, 
# you can simply use the filename when loading the data—there’s no need to specify the file path.

file_path = "Datasets_for_practice/Staff_Salaries.csv"
try:
    df=pd.read_csv(file_path, header=0, encoding='utf-8')
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")
```

What are the columns names?

```python
# Display all columns names
column_names = df.columns
for column in column_names:
    print(column)
```

What are data types of columns we have in this data frame?

```python
# Display data types of all columns
df.dtypes
```

How many rows and columns does this dataframe have?

```python
# Display the number of rows and columns of the dataframe
dimensions = df.shape
numRows = dimensions[0]
numCols = dimensions[1]
print(f"Number of rows: {numRows}")
print(f"Number of columns: {numCols}")
```

How many academic staff are in each discipline?

```python
staff_positions = df['discipline'].value_counts()
staff_positions
```

Create a new variable called selected_staff that includes only the position, gender, and salary columns from the dataframe, then display it on the screen.
