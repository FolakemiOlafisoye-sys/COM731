```python
# Import csv library
import csv

#define empty list
data = []

while True:
    # receive a filename or file path from a user: try to open people.csv file
    file_name = "songs.csv"

    # handle exceptions when opening a file and reading heading and data from it
    try:
        # Open csv file and read data from a file
        with open(file_name, encoding='UTF-8') as csv_file:
            csv_reader = csv.reader(csv_file)
        
            # read in and store the headers separately
            headers = next(csv_reader)

             # read and store the data in a list
            for row in csv_reader:
                data.append(row)
            break # Exit the loop if file is read successfully
    except FileNotFoundError:
        print("File not found. Please try again.")

# Display individual header element
print(headers[0])

#Display song titles
for record in data:
    print(record[0])
```

```python
# Choose the name of the file where you want to save the data.
file_name = "user_entries.csv"

# handle exceptions when opening a file for writing, then prompt the user for a value before writing it to the file
try:
    with open(file_name, 'w') as file:
        #write a column name into a csv file
        file.write("Year, Artist, Song Title\n")
        
        # get data from a user 
        for i in range(3):
            year = int(input("Year: "))
            artist = input("Artist: ")
            title = input("Song Title: ")
            
            # write user input in a format of csv - use commas to separate values
            file.write(f"{year}, {artist}, {title}\n")
        
        print("Thanks! Your file has been written.")

except Exception as e:
    print(f"An error occurred: {e}")
```
