Importing pandas with the alias pd.

```python
import pandas as pd
```

Load dataset to a dataframe called df_yelp

```python
#Load dataset to a dataframe called df_yelp
# the Jupyter Notebook is in the same location as your dataset
file_name = "Yelp_data.csv"

try: 
    df_yelp = pd.read_csv(file_name, encoding='utf-8')
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")
```

Display 5 first record of dataframe

```python
# Display 5 first record of dataframe
df_yelp.head()
```

Find the average star rating for each state (You can ignore the number of reviews for each restaurant). Store the result in df_1 variable

Solution1

```python
#Find the average star rating for each state
df_yelp.groupby('state').stars.mean()
```

Solution 2

```python
df_yelp.groupby('state')['stars'].mean()
```

For each state compute the avg number of stars and review_count they got.

```python
#For each state compute the avg number of stars and review_count they got.
df_yelp.groupby('state')[['stars','review_count']].mean()
```

Find the business that have review_count more than 500 reviews for Arizona(AZ) state

Solution 1

```python
# Find the business that have review_count more than 500 reviews for Arizona(AZ) state
df_2 = df_yelp.groupby('state')

az_df = df_2.get_group('AZ')
az_df.loc[az_df.review_count>=500, :]
```

Solution 2

```python
df_az = df_yelp.groupby('state').get_group('AZ')
df_az
```

```python
df_az[df_az['review_count']>500]
```
