## Import pandas and read this csv into a dataframe called purchase_df

```python
#Write your code here

import pandas as pd

# the Jupyter Notebook is in the same location as your dataset
file_name = "Purchase_data.csv"
try: 
    purchase_df = pd.read_csv(file_name, encoding='utf-8')
except FileNotFoundError as e1:
    print(f"File not found - {e1}")
except Exception as e2:
    print(f"Exception - {e2}")
```

## Display 5 first record of dataframe

```python
#Write your code here
purchase_df.head(5)
```

## Compute the average and max price charged.  Store the results in variables called avg_price and max_price

```python
#Write your code here

avg_price = purchase_df.price.mean()
max_price =purchase_df.price.max()

print(f"Average price: £{avg_price:.2f}")
print(f"Maximum price: £{max_price:.2f}")
```

## Compute the average price and average number of items purchased with only a single call to the mean() method. Store these averages in avg_price and avg_num_items.

```python
#Write your code here
avgs = purchase_df[["price", "num_item"]].mean()
avg_price = avgs.loc["price"]
avg_num_items = avgs.loc["num_item"]

print(f"Average price: {avg_price:.2f}")
print(f"Avergae number of items: {avg_num_items:.2f}")
```

## Convert 'invoice_time' column to be datetime format and update the original DataFrame with this change.

```python
purchase_df['invoice_time'] = pd.to_datetime(purchase_df['invoice_time'],format='mixed',dayfirst=True)
purchase_df.head()
```

## Group the purchase transactions by store_id and save the result in a new variable called group_store_df

```python
group_store_df=purchase_df.groupby('store_id')
```
