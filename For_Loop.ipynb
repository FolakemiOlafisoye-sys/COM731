count_down

```python
distance = int(input('How far are we from the cave?'))
print()

for count in range(distance, 0, -1):
    print(f'{count} steps remaining')

print('We have reached the cave!')
```

Lifetime Earning

```python
name = input('Enter your name: ')
earning = int(input('Enter the amount you earn: '))
age = int(input('Enter the age you started working: '))
total_earning = 0
retired_age  = 65
years_of_service = retired_age - age

for i in range(0, years_of_service):
    if i == 0:
        total_earning = total_earning + earning
        print(total_earning)
    else:
        earning = earning + (earning*0.05)
        total_earning = total_earning + earning
        print(total_earning)

print(f'{name} will earn about ${total_earning:,.2f}')
```
